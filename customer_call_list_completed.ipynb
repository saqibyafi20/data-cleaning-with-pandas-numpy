```python
import pandas as pd
```

```python
data = pd.read_excel('Customer Call List.xlsx')
```

```python
data
```

```python
 data=data.drop_duplicates()
```

```python
data
```

```python
data['Last_Name']=data['Last_Name'].str.strip('._/')
```

```python
data
```

```python
data['Phone_Number']=data['Phone_Number'].astype(str).replace('[^a-zA-Z0-9]','',regex=True)
```

```python
data
```

```python
data['Phone_Number']=data['Phone_Number'].apply(lambda x:x[0:3]+'-' + x[3:6]+'-' +x[6:])
```

```python
data
```

```python
data['Phone_Number'].fillna('')
```

```python
data
```

```python
data['Phone_Number']=data['Phone_Number'].str.replace('Na--','')
```

```python
data
```

```python
data[['street','city','pincode']] = data['Address'].str.split(',',expand=True)
```

```python
data
```

```python
data=data.drop(columns=['Address','city','pincode','Not_Useful_Column'])
```

```python
data
```

```python
data['Paying Customer']=data['Paying Customer'].str.replace("Y",'Yes')
```

```python
data
```

```python
data['Do_Not_Contact']=data['Do_Not_Contact'].str.replace("Y",'Yes')
```

```python
data
```

```python
data
```

```python
data['Do_Not_Contact']=data['Do_Not_Contact'].fillna('Yes')
```

```python
data
```

```python
for i in data.index:
  if data.loc[i,'Do_Not_Contact']=='Yes':
    data.drop(i,inplace=True)
```

```python
data
```

```python

for i in data.index:
  if data.loc[i,'Paying Customer']=='No':
    data.drop(i,inplace=True)
```

```python
data
```

```python
for i in data.index:
  if data.loc[i,'Phone_Number']=='nan--':
    data.drop(i,inplace=True)
```
